```python
import pandas as pd
import re
import spacy
from transformers import AutoTokenizer

# Load tokenizer and spaCy model
tokenizer = AutoTokenizer.from_pretrained("xlm-roberta-base")
nlp = spacy.load("en_core_web_sm")

# Load dataset
df = pd.read_csv("pii_dataset.csv")

# Regex patterns
email_pattern = r'\b[\w\.-]+@[\w\.-]+\.\w+\b'
phone_pattern = r'\b(?:\+?\d{1,3}[-.\s]?)?(?:\d[-.\s]?){8,14}\d\b'
address_pattern = [
    r'\b\d{1,4}[\s\-]?(?:[A-Z][a-z]*[\s\-]?)*(?:Street|St\.|Road|Rd\.|Avenue|Ave\.|Lane|Ln\.?|Block|Sector|Nagar|Colony|Vihar|Marg|Enclave|Circle)\b.*?(?=[.,;]|$)',
    r'\b(?:House\s*No\.?|Flat\s*No\.?)\s*\d+[A-Z]?\b.*?(?=[.,;]|$)',
    r'\b(?:Near|Behind|Opposite|Beside|In front of|Adjacent to)\s+[A-Z][a-zA-Z\s&.\-]+\b'
]


name_pattern = [
    r'\bMy name is ([A-Z][a-z]+(?: [A-Z][a-z]+)*)',
    r'\bMy name is ([A-Z]{1,3}(?: [A-Z][a-z]+)+)',
    r'\bI am ([A-Z][a-z]+(?: [A-Z][a-z]+)*)',
    r'\bI am ([A-Z]{1,3}(?: [A-Z][a-z]+)+)',
    r'\bThis is ([A-Z][a-z]+(?: [A-Z][a-z]+)*)',
    r'\bThis is ([A-Z]{1,3}(?: [A-Z][a-z]+)+)'
]


# Detect pii intexes
def detect_pii_indexes(text):
    indexes = []

    # Email
    for match in re.finditer(email_pattern, text):
        indexes.append((match.start(), match.end(), 'EMAIL'))

    # Phone
    for match in re.finditer(phone_pattern, text):
        indexes.append((match.start(), match.end(), 'PHONE'))

    # Address
    for pattern in address_pattern:
      for match in re.finditer(pattern, text):
        indexes.append((match.start(), match.end(), 'ADDRESS'))


    # spaCy Names
    doc = nlp(text)
    for ent in doc.ents:
        if ent.label_ == "PERSON":
            indexes.append((ent.start_char, ent.end_char, 'NAME'))

    # Regex Names
    for pattern in name_pattern:
        for match in re.finditer(pattern, text):
            name_text = match.group(1)
            start = match.start(1)
            end = match.end(1)
            indexes.append((start, end, 'NAME'))

    return indexes

# Label tokens
def label_tokens(text):
    indexes = detect_pii_indexes(text)
    encoding = tokenizer(text, return_offsets_mapping=True, return_tensors='pt', truncation=True)
    tokens = tokenizer.tokenize(text)
    offsets = encoding['offset_mapping'][0].tolist()
    labels = ['O'] * (len(offsets)-1)

    for index_start, index_end, entity in indexes:
        began = False
        for i, (tok_start, tok_end) in enumerate(offsets):
            if tok_end == 0:  # for start and end of sentence necessary for model
                continue
            if tok_start >= index_end:
                break
            if tok_start >= index_start and tok_end <= index_end:
                if not began:
                    labels[i-1] = f'B-{entity}'
                    began = True
                else:
                    labels[i-1] = f'I-{entity}'
    return tokens, labels

# Applying to DataFrame
token_label_data = []
for text in df['text']:
    tokens, labels = label_tokens(str(text))
    token_label_data.append({'tokens': tokens, 'labels': labels})

# Output to csv
df_labels = pd.DataFrame(token_label_data)
df_labels.to_csv("token_labels_fixed.csv", index=False)

print("Saved as token_labels_fixed.csv")
```
